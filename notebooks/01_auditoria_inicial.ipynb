````python
from pathlib import Path
import pandas as pd
import json

ROOT = Path.cwd()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
DATA_RAW = ROOT / 'data' / 'raw'
RESULTS_TABLES = ROOT / 'results' / 'tables'
RESULTS_TABLES.mkdir(parents=True, exist_ok=True)

print('Raiz do projeto:', ROOT.resolve())
print('Diretório de dados:', DATA_RAW.resolve())
````

## 2. Inventariar os arquivos de dados
Esta célula registra tamanho e extensão dos arquivos presentes em `data/raw`. Ela não modifica os dados originais.

````python
rows = []
for path in sorted(DATA_RAW.rglob('*')):
    if path.is_file():
        rows.append({
            'arquivo': str(path.relative_to(ROOT)).replace('\\\\', '/'),
            'extensao': path.suffix.lower() or '(sem extensão)',
            'tamanho_bytes': path.stat().st_size
        })

inventario = pd.DataFrame(rows)
display(inventario)
out = RESULTS_TABLES / 'inventario_arquivos.csv'
inventario.to_csv(out, index=False, encoding='utf-8-sig')
print(f'Inventário salvo em: {out.relative_to(ROOT)}')
````

## 3. Inspecionar cabeçalhos de arquivos tabulares
A inspeção é exploratória: não define ainda qual dataset será usado para treinar os modelos.

````python
csv_files = list(DATA_RAW.rglob('*.csv'))
for file in csv_files:
    print(f'\\nArquivo: {file.relative_to(ROOT)}')
    try:
        df = pd.read_csv(file, nrows=5)
        print('Colunas:', list(df.columns))
        display(df.head())
    except Exception as exc:
        print('Não foi possível ler automaticamente:', exc)
````
